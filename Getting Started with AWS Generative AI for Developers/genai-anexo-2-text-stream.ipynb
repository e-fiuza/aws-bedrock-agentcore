```python
import boto3
import json
from IPython.display import JSON
```

```python
MODEL_ID = 'amazon.nova-micro-v1:0'
bedrock = boto3.client(service_name='bedrock-runtime', region_name="us-east-1")
```

```python
payload = {
    "messages": [
        {
            "role": "user",
            "content": [
                {
                    "text": "Me diga quais tipos de dança as pessoas praticam."
                }
            ]
        }
    ]
}

payload_json = json.dumps(payload)
```

```python
response = bedrock.invoke_model_with_response_stream(
    modelId=MODEL_ID,
    body=payload_json,
    contentType='application/json',
    accept='application/json'
)
```

```python
stream = response['body']
full_text = ""

for event in stream:
    chunk = event['chunk']
    chunk_str = chunk.get('bytes', b'').decode('utf-8')

    try:
        json_chunk = json.loads(chunk_str)

        if "contentBlockDelta" in json_chunk:
            delta = json_chunk['contentBlockDelta']['delta']
            text = delta.get("text", "")
            full_text += text
            print(text, end='')
    
    except json.JKSONDecodeError:
        continue
```
