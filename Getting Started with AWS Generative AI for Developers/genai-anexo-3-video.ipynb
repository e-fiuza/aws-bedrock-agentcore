```python
import boto3
import json
import random
```

```python
bedrock_runtime = boto3.client("bedrock-runtime", region_name='us-east-1')
bedrock = boto3.client("bedrock", region_name='us-east-1')

s3 = boto3.client("s3")
model_id = "amazon.nova-reel-v1:0"

seed = random.randint(0, 2147483646)
```

```python
prompt="Uma pessoa dançando em uma montanha"
```

```python
model_input = {
    "taskType": "TEXT_VIDEO",
    "textToVideoParams": {"text": prompt},
    "videoGenerationConfig": {
        "fps": 24,
        "durationSeconds": 6,
        "dimension": "1280x720",
        "seed": seed,
    }
}
```

```python
output_config = {
    "s3OutputDataConfig": {
        "s3Uri": "s3://gen-ai-exercise-ef/VIDEO/"
    }
}
```

```python
response = bedrock_runtime.start_async_invoke(
    modelId=model_id,
    modelInput=model_input,
    outputDataConfig=output_config
)
```

```python
invocation_arn = response["invocationArn"]
print("✅ Job enviado!")
print("ARN da invocação:", invocation_arn)
```

```python
job_status = bedrock_runtime.get_async_invoke(invocationArn=invocation_arn)
print("Status atual:", job_status["status"])
```
