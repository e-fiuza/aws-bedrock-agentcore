```python
import boto3
import json
```

```python
bedrock = boto3.client('bedrock', region_name='us-east-1')

models = [
    {"modelArn": "arn:aws:bedrock:us-east-1::foundation-model/amazon.nova-pro-v1:0"},
    {"modelArn": "arn:aws:bedrock:us-east-1::foundation-model/amazon.nova-lite-v1:0"}
]
```

```python
routing_criteria = {"responseQualityDifference": 10} # << Muda de modelo se a diferença de qualidade das respostas passar de 10%

response = bedrock.create_prompt_router(
    promptRouterName="CustomerInquiryRouter-1",
    models=models,
    routingCriteria=routing_criteria,
    fallbackModel=models[0]
)
```

```python
print(json.dumps(response, indent=2))
```
